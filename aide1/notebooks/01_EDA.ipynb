# 01. Exploratory Data Analysis (EDA)

Iris dataset — structure, distributions, and relationships.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["target"] = iris.target
df["species"] = df["target"].map({0: "setosa", 1: "versicolor", 2: "virginica"})
df.head()
```

```python
print("Shape:", df.shape)
print("\nDescribe:")
display(df.describe())
print("\nMissing values per column:")
print(df.isna().sum())
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
df["species"].value_counts().sort_index().plot(kind="bar", ax=ax, color="steelblue", edgecolor="black")
ax.set_title("Target distribution (species)")
ax.set_xlabel("Species")
ax.set_ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
feature_cols = [c for c in df.columns if c not in ("target", "species")]
df[feature_cols].hist(figsize=(10, 8), bins=15, edgecolor="black")
plt.suptitle("Feature distributions", y=1.02)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.ravel()
for i, col in enumerate(feature_cols):
    sns.boxplot(data=df, x="species", y=col, ax=axes[i], palette="Set2")
    axes[i].set_title(col)
plt.suptitle("Box plots by species", y=1.02)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
corr = df[feature_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation heatmap (features)")
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df, hue="species", vars=feature_cols, palette="Set2", corner=True)
plt.suptitle("Pair plot (features by species)", y=1.02)
plt.show()
```

## Summary — key findings

- **Classes**: Three balanced species (50 samples each in the raw sklearn Iris set).
- **Features**: Sepal/petal length and width; petal measures tend to separate *setosa* clearly from the others.
- **Correlations**: Petal length and petal width are strongly correlated; sepal length correlates moderately with petal features.
- **Missing data**: Typically none in the canonical Iris dataset; verify with `isna()` above.
- **Next step**: Save raw data for reproducible preprocessing and modeling.
